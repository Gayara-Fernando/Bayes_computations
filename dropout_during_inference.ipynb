```python
import tensorflow as tf
import numpy as np
```

```python
# let's generate some data
```

```python
y = np.random.normal(5,2,100)
```

```python
X = np.concatenate((np.random.normal(0,1,100).reshape(-1,1), np.random.random(100).reshape(-1,1)), axis = 1)
```

```python
X.shape
```

```python
input_layer = tf.keras.Input(shape = 2)
hidden_layer = tf.keras.layers.Dense(6)
hidden_output = hidden_layer(input_layer)
dropout = tf.keras.layers.Dropout(0.5)
dropped_output = dropout(hidden_output)
pred_layer = tf.keras.layers.Dense(1)
pred_output = pred_layer(dropped_output)
```

```python
model = tf.keras.models.Model(input_layer, pred_output)
```

```python
model.summary()
```

```python
# compile the model
model.compile(optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.001), loss = 'mean_squared_error', metrics = ['mae'])
```

```python
history = model.fit(X, y, epochs = 20)
```

```python
new_x = np.array([[2.5, 0.1]])
```

```python
new_x.shape
```

```python
model.predict(new_x)
```

```python
weights = model.get_weights()
```

```python
len(weights)
```

```python
weights[0].shape, weights[1].shape, weights[2].shape, weights[3].shape
```

```python
def forward_pass(x, W1, b1, W2, b2):
    # Hidden layer
    h = np.dot(W1.T,x.T) + b1

    # Output layer
    y_hat = W2.T@h + b2 # Sigmoid activation for output

    return y_hat
```

```python
h_dims = forward_pass(new_x, weights[0], weights[1].reshape(-1,1), weights[2], weights[3])
    
```
